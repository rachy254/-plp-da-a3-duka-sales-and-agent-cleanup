## First-Five-Minutes Inspection & Derived Columns

```
import numpy as np
import pandas as pd

df = pd.read_csv('duka_sales.csv')
df['order_date'] = pd.to_datetime(df['order_date'])

print('Shape:', df.shape)
print('\nInfo:')
print(df.dtypes)
print('\nMissing values:\n', df.isna().sum())
print('\nDate range:', df['order_date'].min(), 'to', df['order_date'].max())
display(df.head())
display(df.describe(include='all'))

df['revenue'] = df['unit_price'] * df['quantity']
df['month'] = df['order_date'].dt.month_name()
df['weekday'] = df['order_date'].dt.day_name()
df['is_weekend'] = df['order_date'].dt.dayofweek >= 5
df['price_band'] = np.where(df['unit_price'] >= 28000, 'Premium', 'Value')
display(df[['order_id', 'revenue', 'month', 'weekday', 'is_weekend', 'price_band']].head())
```

## Task 3: Branch Revenue Ranking

```
branch_summary = (
    df.groupby('branch')
    .agg(total_revenue=('revenue', 'sum'), orders=('order_id', 'count'), avg_order_value=('revenue', 'mean'))
    .sort_values('total_revenue', ascending=False)
)
display(branch_summary)
```

### Answer — Task 3 (Branch Revenue Ranking)

**Nairobi** ranks first in revenue at **KES 14,901,737.00** across **313** orders, followed by **Kisumu** (**KES 8,081,583.00**). Lowest revenue is in **Eldoret** (**KES 5,774,810.00**).

## Task 4: Product & Price Band Breakdown

```
product_summary = (
    df.groupby(['product', 'price_band'])
    .agg(total_revenue=('revenue', 'sum'), orders=('order_id', 'count'), avg_unit_price=('unit_price', 'mean'))
    .sort_values('total_revenue', ascending=False)
)
display(product_summary)
```

### Answer — Task 4 (Product & Price Band Performance)

**Smartphone** generates the highest total revenue (**KES 20,103,344.00**), while **355** orders fall into the **Premium** price band (`unit_price >= 28,000`) versus **645** orders in the **Value** band.

## Task 5: Weekend vs. Weekday Performance

```
weekend_summary = df.groupby('is_weekend').agg(
    total_revenue=('revenue', 'sum'),
    orders=('order_id', 'count'),
    avg_order_value=('revenue', 'mean'),
)
display(weekend_summary)
```

### Answer — Task 5 (Weekend vs. Weekday Comparison)

Weekday orders generated **KES 30,396,172.00** across **702** orders, compared to **KES 12,534,230.00** across **298** weekend orders. Average order value is **KES 42,061.17** on weekends vs. **KES 43,299.39** on weekdays.

## Task 6: Monthly Revenue Trend (January–June 2026)

```
months = ['January', 'February', 'March', 'April', 'May', 'June']
monthly_summary = df.groupby('month').agg(total_revenue=('revenue', 'sum'), orders=('order_id', 'count')).reindex(months)
display(monthly_summary)
```

### Answer — Task 6 (Monthly Revenue Trend, Jan–Jun 2026)

Monthly revenue peaked in **March** (**KES 8,437,460.00**) and was lowest in **May** (**KES 5,622,380.00**).

## Task 7: Filtered Export (`outputs/premium_online_coast_capital.csv`)

```
import os
os.makedirs('outputs', exist_ok=True)
filtered = df[
    (df['price_band'] == 'Premium')
    & (df['channel'] == 'Online')
    & (df['branch'].isin(['Nairobi', 'Mombasa']))
]
filtered.to_csv('outputs/premium_online_coast_capital.csv', index=False)
print('Saved rows:', len(filtered))
display(filtered.head())
```

### Answer — Task 7 (Filtered Export: Premium, Online, Nairobi & Mombasa)

Filtering for `price_band == 'Premium'`, `channel == 'Online'`, and `branch in ['Nairobi', 'Mombasa']` yielded **81** rows totaling **KES 8,231,316.00**, saved to **`outputs/premium_online_coast_capital.csv`** with `index=False`.
